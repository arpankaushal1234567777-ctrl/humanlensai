# HumanLens AI — Research Questions & Hypotheses

## Research / ML Lead

HumanLens AI is a multimodal human behavior analysis and de-escalation
system designed to analyze observable communication and behavioral signals,
estimate potential communication escalation, and provide users with an
opportunity to pause, reflect, and respond more constructively.

This notebook defines the research questions, measurable hypotheses,
research objectives, variables, evaluation metrics, and experimental
direction for the Research / ML component of HumanLens AI.

# 1. Research Problem

Human communication can contain observable signals associated with
escalation, such as aggressive language, toxic language, insults,
threats, increased emotional intensity, and changes in behavioral context.

The central research problem of HumanLens AI is:

> Can observable text, voice, facial-expression, and longitudinal
> behavioral signals be combined to estimate potential communication
> escalation and provide an explainable opportunity for de-escalation?

The system should focus on communication and behavioral context rather than
making medical or psychological diagnoses.

The research will investigate whether combining multiple structured signals
provides useful improvements over text-only analysis while maintaining
interpretability, reasonable latency, privacy, and fairness.

# 2. Research Objectives

The main objectives of the Research / ML component are:

1. Develop a reliable text-based system for toxicity and aggression
   analysis.

2. Develop an emotion and emotional-intensity analysis layer.

3. Develop behavioral features that represent longitudinal patterns
   rather than judging a user from a single interaction.

4. Investigate voice features such as speaking rate, pitch, pauses,
   and intensity.

5. Investigate broad facial-expression signals without facial recognition
   or identity tracking.

6. Develop an explainable multimodal fusion method.

7. Determine whether multimodal signals improve escalation detection
   compared with text-only analysis.

8. Evaluate model performance using precision, recall, F1-score,
   confusion matrices, calibration, and latency.

9. Analyze false positives, false negatives, uncertainty, and potential
   demographic bias.

10. Evaluate whether a pause/rewrite intervention can reduce the toxicity
    or aggression of a message.

# 3. Research Questions

## RQ1 — Text Analysis

Can text-based analysis detect aggressive or escalatory communication
with useful precision and recall?



## RQ2 — Multimodal Improvement

Does adding voice and facial-expression signals improve escalation
detection compared with text-only analysis?



## RQ3 — Behavioral Context

Which behavioral factors are associated with higher self-reported
irritability or escalation events?



## RQ4 — Intervention Effectiveness

Does a short pause/rewrite intervention reduce message toxicity
or aggression?



## RQ5 — Context and Demographic Variation

Are observed patterns consistent across different age groups, or are
they more strongly explained by behavioral and contextual factors?

# 4. Research Hypotheses

## H1 — Text Classification

A trained text classification model can identify toxic and aggressive
communication with useful precision, recall, and F1-score on a held-out
evaluation dataset.

### Measurement

- Precision
- Recall
- F1-score
- Confusion matrix
- Calibration
- Inference latency


## H2 — Multimodal Fusion

A multimodal model combining text with voice and/or facial-expression
signals will provide equal or better escalation detection performance
than a text-only model.

### Comparison

1. Text only
2. Text + Voice
3. Text + Voice + Face
4. Text + Voice + Face + Behavioral Context

### Measurement

- Precision
- Recall
- F1-score
- Calibration
- False-positive rate
- False-negative rate


## H3 — Behavioral Association

Longitudinal behavioral features such as stress, sleep stability,
academic pressure, mood, social interaction, and activity will show
measurable associations with self-reported irritability or escalation
events.

This hypothesis concerns association, not causation.


## H4 — De-escalation Intervention

A pause/rewrite intervention can reduce the measured toxicity,
aggression, or emotional intensity of a message compared with the
original message.


## H5 — Fairness and Context

Model performance may vary across available demographic groups, but
behavioral and contextual features should not be replaced by demographic
group membership as a direct basis for predicting escalation.

# 6. Model Comparison Strategy

The research will not assume that a more complex model is automatically
better.

The following progression will be evaluated:

### Baseline 1 — Rule-Based

Simple deterministic indicators such as:

- Threat-related terms
- Insult-related terms
- Excessive punctuation
- Capitalization patterns
- Explicit toxic terms

### Baseline 2 — Classical Machine Learning

- TF-IDF + Logistic Regression
- TF-IDF + Linear SVM

### Model 3 — Pre-trained NLP Model

A suitable pre-trained transformer-based model will be evaluated.

### Multimodal Models

- Text only
- Text + Voice
- Text + Face
- Text + Voice + Face
- Text + Voice + Face + Behavioral Context

The first multimodal fusion approach will use explainable weighted
or learned structured scoring rather than an opaque end-to-end model.

# 7. Evaluation Metrics

The following metrics will be used throughout the research.

| Metric | Purpose |
|---|---|
| Precision | Measures how many predicted positive cases are actually positive |
| Recall | Measures how many relevant positive cases are detected |
| F1-score | Balances precision and recall |
| Confusion Matrix | Identifies false positives and false negatives |
| Calibration | Evaluates whether confidence scores are meaningful |
| Latency | Measures suitability for real-time interaction |
| False Positive Rate | Measures unnecessary interventions |
| False Negative Rate | Measures missed escalation cases |

For intervention experiments, additional measurements include:

- Percentage of users choosing pause/rewrite
- Change in toxicity/aggression
- Change in emotional intensity
- User-rated usefulness
- Whether users feel supported or judged

# 8. Research Constraints and Ethical Principles

HumanLens AI will follow the following principles:

1. The system is not a medical diagnostic tool.

2. Behavioral scores must not be presented as diagnoses.

3. Demographic information must not be used as a simple proxy for
   behavioral risk.

4. Models must be evaluated for false positives and false negatives.

5. Uncertainty should be represented where appropriate.

6. Political criticism, disagreement, or profanity should not
   automatically be treated as harmful escalation.

7. Facial analysis must not perform identity recognition or tracking.

8. Raw audio/video should not be stored by default.

9. Model versions and analysis versions should be recorded for
   reproducibility.

10. Research involving users or controlled intervention experiments
    requires appropriate consent and review.
